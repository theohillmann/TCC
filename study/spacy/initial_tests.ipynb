```python
import spacy
from spacy import displacy
import numpy as np
```

```python
npl = spacy.load('en_core_web_sm')
```

```python
with open('sample_text.txt', 'r') as f:
    sample_text = f.read()
```

```python
doc = npl(sample_text)
```

```python
for token in doc[:10]:
    print(token)
```

```python
for sentence in doc.sents:
    print(sentence)
```

```python
token2 = list(doc.sents)[1][1]
token2
```

```python
print(token2.left_edge, token2, token2.right_edge)
print(token2, token2.lemma_)
print(token2.morph)
print(token2.lang_)
```

```python
nlp = spacy.load('en_core_web_md')
```

```python
doc = nlp(sample_text)
sentence1 = list(doc.sents)[0]
print(sentence1)
```

```python
word = 'dizzy'
ms = nlp.vocab.vectors.most_similar(
    np.asarray([nlp.vocab.vectors[nlp.vocab.strings[word]]]), n=10)
words = [nlp.vocab.strings[w] for w in ms[0][0]]
distances = ms[2]
print(words)
```
